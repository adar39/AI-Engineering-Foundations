```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x = np.linspace(0,5,11)
x
```

```python
y = x ** 2
y
```

```python
plt.plot(x,y)
```

```python
y = x ** 3
plt.plot(x,y)
plt.title("Cubic Graph")
plt.xlabel("X")
plt.ylabel("Y")
```

**Subplots in Matplotlib**

```python
# Function Syntax: plt.subplot(nrows, ncols, index)

plt.subplot(2,2,1)
plt.subplot(2,2,4)

plt.subplot(2,2,3)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(y,x)
```

**Object Oriented way**

```python
# Basic syntax of a Completely custom pllotting area : fig.add_axes([left, bottom, width, height])

fig = plt.figure()

axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x,y)

axis2 = fig.add_axes([0.5, 0.5 , 0.4, 0.8])
axis2.plot(x,y)
```

```python
'''
 Syntax: plt.figure(num=None, figsize=None, dpi=None, facecolor=None, edgecolor=None, frameon=True)
 num: Sets figure ID/window title | figsize: (width, height) in inches | dpi: Pixels per inch
 facecolor: Background color | edgecolor: Border color | frameon: Draws background frame if True
'''

fig = plt.figure(figsize=(20,8), dpi = 100)

axis1 = fig.add_axes([0.1, 0.1, 1, 1])
axis1.plot(x,y)

axis2 = fig.add_axes([0.3, 0.4 , 0.3, 0.8])
axis2.plot(x,y)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# FIX 1: Initialize the missing x-axis data
xaxis = np.linspace(0, 5, 11)

fig, ax = plt.subplots(figsize=(12,6))

# Line widths
ax.plot(xaxis, xaxis+1, color="red", linewidth=0.25)
ax.plot(xaxis, xaxis+2, color="red", linewidth=0.50)
ax.plot(xaxis, xaxis+3, color="red", linewidth=1.00)
ax.plot(xaxis, xaxis+4, color="red", linewidth=2.00)

# FIX 2: Fixed comment typo (en-dash replaced with standard hyphen)
# possible linestyle options '-', '--', '-.', ':'
ax.plot(xaxis, xaxis+5, color="green", lw=3, linestyle='-')
ax.plot(xaxis, xaxis+6, color="green", lw=3, ls='-.')
ax.plot(xaxis, xaxis+7, color="green", lw=3, ls=':')

# FIX 3: Replaced invalid linestyle='steps' with the correct function if needed
# ax.step(xaxis, xaxis+5.5, color="green", lw=3) # Use this if you actually want a step chart

# Custom dash
line, = ax.plot(xaxis, xaxis+8, color="black", lw=1.50)
line.set_dashes([5, 50, 15, 50]) 

# Markers
ax.plot(xaxis, xaxis+ 9, color="blue", lw=3, ls='-', marker='+')
ax.plot(xaxis, xaxis+10, color="blue", lw=3, ls='--', marker='o')
ax.plot(xaxis, xaxis+11, color="blue", lw=3, ls='-', marker='s')
ax.plot(xaxis, xaxis+12, color="blue", lw=3, ls='--', marker='1')

# Marker sizing and coloring
ax.plot(xaxis, xaxis+13, color="purple", lw=1, ls='-', marker='o', markersize=2)
ax.plot(xaxis, xaxis+14, color="purple", lw=1, ls='-', marker='o', markersize=4)
ax.plot(xaxis, xaxis+15, color="purple", lw=1, ls='-', marker='o', markersize=8, markerfacecolor="red")
ax.plot(xaxis, xaxis+16, color="purple", lw=1, ls='-', marker='s', markersize=8, 
        markerfacecolor="yellow", markeredgewidth=3, markeredgecolor="green")

plt.show()
```
